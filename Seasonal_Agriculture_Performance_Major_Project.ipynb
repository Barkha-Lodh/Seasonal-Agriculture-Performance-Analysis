# Outlier Analysis

```python
# Outlier Analysis using IQR

Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q2 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q2 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q2 + 1.5 * IQR

yield_outliers = df[
    (df['Yield_Tonnes_Ha'] < lower_limit) |
    (df['Yield_Tonnes_Ha'] > upper_limit)
]

print("Q1:", Q1)
print("Q2:", Q2)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Yield Outliers:", len(yield_outliers))
```

# Multivariate Analysis
## Pairplot

```python
# Multivariate Analysis using Pairplot

selected_columns = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Farm_Area_Hectares',
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Profit_INR'
]

multivariate_data = df[selected_columns].sample(
    500,
    random_state=42
)

sns.pairplot(multivariate_data)

plt.show()
```

```python
# Select numerical columns

numerical_data = df.select_dtypes(include=np.number)

correlation = numerical_data.corr()

plt.figure(figsize=(16,12))
sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='Purples',
    center=0
)

plt.title("Correction Matrix of Agriculture Variables")
plt.show()
```

```python
#Yield Correlation

yield_correlation = correlation['Yield_Tonnes_Ha'].sort_values(ascending=False)
display(yield_correlation)
```

```python
from scipy.stats import f_oneway

season_groups = [
    group['Yield_Tonnes_Ha'].dropna()
    for name, group in df.groupby('Season')
]

anova_stat, p_value = f_oneway(*season_groups)

print("ANOVA Statistic:", anova_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Yield differs significantly across seasons.")
else:
    print("Result: No statistically significant difference in yield across seasons.")
```
